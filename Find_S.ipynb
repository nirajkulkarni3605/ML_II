```python


data = [
    ["Mountain", "Clear", "Patrol", "High", "Advanced", "Yes"],
    ["Mountain", "Clear", "Patrol", "High", "Basic", "Yes"],
    ["Mountain", "Rainy", "Patrol", "High", "Advanced", "Yes"],
    ["Desert", "Clear", "Patrol", "High", "Advanced", "No"],
    ["Mountain", "Clear", "Rescue", "High", "Advanced", "No"],
    ["Mountain", "Clear", "Patrol", "Low", "Advanced", "No"]
]


num_attributes = len(data[0]) - 1


hypothesis = ["∅"] * num_attributes

print("Find-S Algorithm")



for i, example in enumerate(data):

   
    if example[-1] == "Yes":

        
        if hypothesis == ["∅"] * num_attributes:
            hypothesis = example[:-1].copy()

        
        else:
            for j in range(num_attributes):
                if hypothesis[j] != example[j]:
                    hypothesis[j] = "?"

        print(f"After positive example {i + 1}: {hypothesis}")


print("\nFinal Hypothesis:")
print(hypothesis)
```
